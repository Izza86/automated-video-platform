# 🚀 Google Colab GPU Setup (Connect VS Code to Colab)

## Step 1: Open Google Colab
1. Go to [Google Colab](https://colab.research.google.com/)
2. Create a **new notebook**
3. Go to **Runtime → Change runtime type → GPU (T4)** → Save

## Step 2: Run this code **IN COLAB** (not here)
Copy and paste the code from **Cell 3** below into your Colab notebook and run it there.  
It will output a URL like `http://localhost:8888/?token=...` — copy that URL.

## Step 3: Connect VS Code to Colab
1. In VS Code, press `Ctrl+Shift+P`
2. Search **"Jupyter: Specify Jupyter Server for Connections"**
3. Select **"Existing"**
4. Paste the URL from Colab
5. Select the remote kernel for this notebook (bottom-right kernel picker)

## Step 4: Verify GPU
Run **Cell 4** (`!nvidia-smi`) to confirm the T4 GPU is available.

## ⬇️ Copy the code in Cell 3 below and paste it into your Google Colab notebook
**Do NOT run it here** — it must be run inside Google Colab to expose the GPU runtime.

```python
# ============================================================
# 📋 COPY THIS ENTIRE CELL INTO GOOGLE COLAB AND RUN IT THERE
# ============================================================

# Step 1: Install jupyter_http_over_ws extension
!pip install jupyter_http_over_ws
!jupyter serverextension enable --py jupyter_http_over_ws

# Step 2: Start Jupyter server (this will output a URL — copy it!)
!jupyter notebook \
  --NotebookApp.allow_origin='https://colab.research.google.com' \
  --port=8888 \
  --NotebookApp.port_retries=0 \
  --no-browser \
  --allow-root

# After running, look for a line like:
#   http://localhost:8888/?token=abc123...
# Copy that FULL URL and paste it into VS Code (Step 3 in Cell 1)
```

```python
# ✅ Run this cell AFTER connecting to Colab's GPU kernel
# It should show a Tesla T4 GPU if connected correctly
!nvidia-smi
```

```python
# ✅ Additional GPU check — verifies PyTorch can see the GPU
import torch
print(f"PyTorch version: {torch.__version__}")
print(f"CUDA available: {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"GPU Name: {torch.cuda.get_device_name(0)}")
    print(f"GPU Memory: {torch.cuda.get_device_properties(0).total_mem / 1e9:.1f} GB")
else:
    print("⚠️ No GPU detected — make sure you're connected to Colab's GPU kernel!")
```
